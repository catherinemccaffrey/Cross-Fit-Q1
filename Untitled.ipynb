IMPORTING CROSSFIT FORMULAS

```python
def load_photometry_from_catalog(row):
    """
    Load Euclid VIS/Y/J/H photometry from one row
    of the EW1000 catalog.
    """

    band_names = [
        "Euclid_VIS.vis",
        "Euclid_NISP.Y",
        "Euclid_NISP.J",
        "Euclid_NISP.H",
    ]

    flux_columns = [
        "flux_vis_2fwhm_aper",
        "flux_y_2fwhm_aper",
        "flux_j_2fwhm_aper",
        "flux_h_2fwhm_aper",
    ]

    error_columns = [
        "fluxerr_vis_2fwhm_aper",
        "fluxerr_y_2fwhm_aper",
        "fluxerr_j_2fwhm_aper",
        "fluxerr_h_2fwhm_aper",
    ]

    fluxes = []
    errors = []
    valid_bands = []

    for band, flux_col, err_col in zip(
        band_names, flux_columns, error_columns
    ):
        flux = float(row[flux_col])
        error = float(row[err_col])

        if np.isfinite(flux) and np.isfinite(error) and flux > 0 and error > 0:
            valid_bands.append(band)
            fluxes.append(flux)
            errors.append(error)

    filt = FilterSet.from_names(valid_bands)

    phot = PhotometryData(
        flux=np.array(fluxes),
        error=np.array(errors)
    )

    return filt, phot
```

```python
templates = TemplateLibrary.from_eazy_param_file(
    "templates/fsps_full/tweak_fsps_QSF_12_v3.param"
)

zgrid = np.arange(0.05, 4.0, 0.02)

config = JointFitConfig(
    zgrid=zgrid,
    w_spec=1,
    polynomial_order=0,
    resolution_R=450.0,
    velocity_fwhm=100.0,
    spec_sys_err=0.02,
    phot_min_err=0.02
)

fitter = JointTemplateFitter(config)
phot_projector = PhotometryProjector()
```

```python
def compute_sed(z, coeffs, wave_grid):
    sed = np.zeros_like(wave_grid)
    for comp, c in zip(templates.components, coeffs):
        if c == 0:
            continue
        obs_wave = comp.wave_rest * (1.0 + z)
        obs_flux = comp.eazy_template.flux_flam(z=z)
        sed += c * np.interp(wave_grid, obs_wave, obs_flux, left=0.0, right=0.0)
    return sed
```

RUNNING THE JOINT FIT

```python
import os

fit_results = []

for i, path in enumerate(spectra):

    # -----------------------------
    # 1. Get object ID from filename
    # -----------------------------
    filename = os.path.basename(path)

    source_id = int(
        filename.replace("SPECTRA_RGS-sedm ", "")
                .replace(".fits", "")
    )

    # -----------------------------
    # 2. Find matching catalog row
    # -----------------------------
    matches = ew1000[ew1000["object_id"] == source_id]

    if len(matches) == 0:
        print(f"No catalog match found for {source_id}")
        continue

    row = matches[0]

    print(f"Fitting object {source_id}")

    # -----------------------------
    # 3. Load photometry
    # -----------------------------
    filt, phot = load_photometry_from_catalog(row)

    # -----------------------------
    # 4. Load RGS spectrum
    # -----------------------------
    obj_table = Table.read(path)
    spec = spectrum_from_euclid_fits(obj_table)

    # -----------------------------
    # 5. Photometry-only fit
    # -----------------------------
    phot_only_spectrum = replace(
        spec,
        mask=np.zeros_like(spec.mask, dtype=bool)
    )

    phot_only_data = JointFitInput(
        photometry=phot,
        spectrum=phot_only_spectrum
    )

    result_phot_only = fitter.fit(
        data=phot_only_data,
        templates=templates,
        filters=filt
    )

    # -----------------------------
    # 6. Calibrate spectrum
    # -----------------------------
    cal = scale_spectrum_to_photometry(
        spec,
        phot,
        filt,
        scale_poly_order=1
    )

    calibrated_spectrum = cal.spectrum

    # -----------------------------
    # 7. Joint fit
    # -----------------------------
    # Use the full spectrum, except pixels with invalid flux/error
    good_pixels = (
        np.isfinite(calibrated_spectrum.flux) &
        np.isfinite(calibrated_spectrum.error) &
        (calibrated_spectrum.error > 0)
        )

    joint_spectrum = replace(
        calibrated_spectrum,
        mask=~good_pixels
        )

    print(
    f"Object {source_id}: "
    f"{np.sum(good_pixels)}/{len(good_pixels)} pixels used"
        )

    result_joint = fitter.fit(
        data=JointFitInput(
            photometry=phot,
            spectrum=joint_spectrum
        ),
        templates=templates,
        filters=filt
    )

    # -----------------------------
    # 8. Wavelength grid
    # -----------------------------
    wide_wave = np.linspace(
        0.9 * min(
            spec.wavelength.min(),
            filt.pivot_wavelengths.min()
        ),
        1.1 * max(
            spec.wavelength.max(),
            filt.pivot_wavelengths.max()
        ),
        2000
    )

    # -----------------------------
    # 9. Compute SEDs
    # -----------------------------
    sed_phot_only = compute_sed(
        result_phot_only.zbest,
        result_phot_only.coeffs,
        wide_wave
    )

    sed_joint = compute_sed(
        result_joint.zbest,
        result_joint.coeffs,
        wide_wave
    )

    # -----------------------------
    # 10. Save results
    # -----------------------------
    fit_results.append({
        "object_id": source_id,
        "path": path,
        "photometry": phot,
        "filters": filt,
        "spectrum": joint_spectrum,
        "phot_only": result_phot_only,
        "joint": result_joint,
        "wide_wave": wide_wave,
        "sed_phot_only": sed_phot_only,
        "sed_joint": sed_joint
    })

print(f"Completed {len(fit_results)} fits.")
```

PLOTTING THE RESULTS

```python
for r in fit_results:

    filt = r["filters"]
    phot = r["photometry"]
    spec = r["spectrum"]

    res_joint = r["joint"]
    res_phot = r["phot_only"]

    fig, axes = plt.subplots(
        1, 2,
        figsize=(13, 4.5),
        gridspec_kw={"width_ratios": [3, 1]}
    )

    # ====================================
    # Left panel
    # ====================================

    ax = axes[0]

    finite_err = np.isfinite(spec.error)

    err_lo = np.where(
        finite_err,
        np.clip(spec.flux - spec.error, 1e-21, None),
        np.nan
    )

    err_hi = np.where(
        finite_err,
        spec.flux + spec.error,
        np.nan
    )

    ax.fill_between(
        spec.wavelength,
        err_lo,
        err_hi,
        color="lightsteelblue",
        alpha=0.20,
        linewidth=0,
        label="spectrum ±1σ"
    )

    ax.scatter(
        spec.wavelength,
        spec.flux,
        s=10,
        color="tab:blue",
        alpha=0.7,
        edgecolor="none",
        label="spectrum (calibrated)"
    )

    ax.errorbar(
        filt.pivot_wavelengths,
        phot.flux,
        yerr=phot.error,
        fmt="o",
        color="goldenrod",
        markeredgecolor="black",
        markeredgewidth=0.8,
        markersize=12,
        capsize=3,
        zorder=10,
        label="photometry"
    )

    ax.plot(
        r["wide_wave"],
        r["sed_joint"],
        color="mediumseagreen",
        lw=2,
        label=f"joint best-fit (z={res_joint.zbest:.3f})"
    )

    ax.plot(
        r["wide_wave"],
        r["sed_phot_only"],
        color="chocolate",
        linestyle="--",
        lw=2,
        label=f"photometry-only best-fit (z={res_phot.zbest:.3f})"
    )

    #ax.set_yscale("log")

    ax.set_xlabel("observed wavelength [Å]")
    ax.set_ylabel(r"f$_\lambda$ [erg/s/cm2/Å]")

    ax.set_title(r["path"].split("/")[-1])

    ax.legend(
        frameon=False,
        fontsize=9,
        loc="upper right"
    )

    # ====================================
    # Right panel
    # ====================================

    ax = axes[1]

    dchi2_joint = (
        res_joint.chi2_total
        - res_joint.chi2_total.min()
    )

    dchi2_phot = (
        res_phot.chi2_total
        - res_phot.chi2_total.min()
    )

    ax.plot(
        res_joint.zgrid,
        dchi2_joint,
        color="mediumseagreen",
        lw=2,
        label="joint fit"
    )

    ax.axvline(
        res_joint.zbest,
        color="mediumseagreen",
        linestyle=":",
        lw=2,
        label=f"joint zbest = {res_joint.zbest:.3f}"
    )

    ax.plot(
        res_phot.zgrid,
        dchi2_phot,
        color="chocolate",
        linestyle="--",
        lw=2,
        label="photometry-only"
    )

    ax.axvline(
        res_phot.zbest,
        color="chocolate",
        linestyle=":",
        lw=2,
        label=f"phot-only zbest = {res_phot.zbest:.3f}"
    )

    ax.set_title("Redshift fit")

    ax.set_xlabel("redshift")
    ax.set_ylabel(r"$\Delta\chi^2$")

    ax.legend(
        frameon=False,
        fontsize=7,
        loc="upper right"
    )

    fig.tight_layout()
    plt.show()
```
